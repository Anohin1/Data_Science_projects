## Предобработка данных

### Определите и изучите пропущенные значения. ###

**Определим процент пропусков по каждому столбцу.**

```python
df.isna().mean()*100 # считаем процент пропусков
```

Нужные нам столбцы с большим количеством пропусков: "ceiling_height" - 39 % пропусков, "days_exposition" - 13 % пропусков, "cityCenters_nearest" - 23 %. Мы  понимаем, что при исследовании где затронуты эти величины, мы теряем 39, 13 , 23 % выборки соответственно. Это может повлиять на результат по отдельным категориям квартир. Поэтому удалять пропуски в этих столбцах или  на что-то менять их я не буду. Отсутствие значения — тоже важный сигнал, который не нужно прятать.

Возможные причины появления пропусков. Из условия мы знаем , что по каждой квартире на продажу доступны два вида данных. Первые вписаны пользователем, вторые — получены автоматически на основе картографических данных. Например, расстояние до центра, аэропорта, ближайшего парка и водоёма. 

пропуски в первом виде данных, полученных от пользователя - человеческий фактор; во втором виде пропуски связаны с тем, я полагаю, что системе не хватило данных или она не смогла правильно распознать данные для дальнейшего автоматического заполнения, или же программный сбой.

Как нам посоветовали в условии, заменим пропуски в столбце "balcony" нулями, полагая, что если ничего не указано - то балкона просто нет.

Пропуски в остальных столбцах нам не критичны, так как в нашем исследовании они не используются. Трогать их не будем.

**Заполню нулями пропущенные значения столбца 'balcony'**

```python
df['balcony'] = df['balcony'].fillna(0)
```

```python
df['balcony'].isna().mean()*100 # проверим, считаем долю пропусков в столбце "балконы"
```

### Приведите данные к нужным типам. ###
**Преобразуем данные в столбце 'balcony' в целочисленный тип.**

```python
df['balcony'] = df['balcony'].astype('int')  
# преобразуем тип значений столбца total_income из float в int
df['balcony'].dtype
```

**Преобразуем в целочисленный тип значения остальных столбцов, где это необходимо.** Неиспользуемые в нашем исследовании столбцы преобразуем в целочисленный тип для удобства восприятия значений в таблице.

```python
df['floors_total'] = df['floors_total'].astype('Int64')  
# преобразуем тип значений столбца floors_total из float в int
```

```python
df['airports_nearest'] = df['airports_nearest'].astype('Int64')
df['cityCenters_nearest'] = df['cityCenters_nearest'].astype('Int64')
df['parks_around3000'] = df['parks_around3000'].astype('Int64')
df['parks_nearest'] = df['parks_nearest'].astype('Int64')
df['ponds_around3000'] = df['ponds_around3000'].astype('Int64')
df['ponds_nearest'] = df['ponds_nearest'].astype('Int64')
df['days_exposition'] = df['days_exposition'].astype('float32').astype('Int64')

df.info()
```

```python
df.duplicated().sum() # проверим есть ли дубликаты
```

```python
df.isna().mean()*100
```

в столбце 'floors_total' обнаружены пропуске 0,3 % от всей выборки, необходимо избавиться от пропусков в данном столбце, чтобы на его основе заполнить пропуски 'ceiling_height'.
Будем полагать, что если этажность не указана, то это одноэтажный дом. Поэтому заменим пропуски 1.

```python
df['floors_total'] = df['floors_total'].fillna(1)
```

Сгруппируем датафрейм по столбцу 'floors_total' - этажность, найдём медианные значения высоты потолков для каждой категории домов и заполним соответствующими значениями пропуски в 'ceiling_height'.

```python
df['ceiling_height'] = df.groupby('floors_total')['ceiling_height'].transform(lambda x : x.fillna(x.median()))
```

```python
df['ceiling_height'].isna().mean()*100 #проверим
```

В одной строке значение не заполнилось, так как нет среднего значения под эту категорию. Заполним средним - 2.65

```python
df['ceiling_height'] = df['ceiling_height'].fillna(2.65)
```
